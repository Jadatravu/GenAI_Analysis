```python
"""
LangGraph version of the earlier Phidata agent, using a ReAct-style agent that can
optionally call the DuckDuckGo tool when needed.

Setup:
  1) Create and activate a virtualenv
     python3 -m venv .venv && source .venv/bin/activate
  2) Create requirements.txt with the block below and install
     pip install -r requirements.txt
  3) Set your LLM key
     export OPENAI_API_KEY=...   # for ChatOpenAI via langchain-openai

Run:
  python lg_web_agent.py --q "what's new with the James Webb telescope?"

requirements.txt (example):
  langgraph>=0.2.35
  langchain>=0.2.16
  langchain-openai>=0.1.22
  langchain-community>=0.2.11
  duckduckgo-search>=6.1.0
  rich>=13.7.0
  requests>=2.31.0
"""

import os
import argparse
from typing import Optional

from langchain_openai import ChatOpenAI
from langchain_community.tools import DuckDuckGoSearchRun
from langgraph.prebuilt import create_react_agent
import time
from dotenv import load_dotenv
try:
    from rich.console import Console
    console = Console()
except Exception:  # pragma: no cover
    class _Dummy:
        def print(self, *a, **k):
            print(*a)
    console = _Dummy()

load_dotenv()
def build_web_agent(model_name: Optional[str] = None):
    """Create a LangGraph ReAct agent with DuckDuckGo as a tool.

    The agent uses tool-calling to decide when to search the web.
    """
    llm_id = model_name or os.getenv("OPENAI_MODEL", "gpt-4o-mini")
    llm = ChatOpenAI(model=llm_id, temperature=0)

    ddg = DuckDuckGoSearchRun()
    tools = [ddg]

    # create_react_agent wires up a small LangGraph under the hood for tool calls
    agent = create_react_agent(llm, tools)
    return agent


def ask_web_agent(query: str, model_name: Optional[str] = None, stream: bool = True) -> None:
    """Send a prompt to the LangGraph agent and print the response.

    If stream=True, we display intermediate tool/agent messages as they occur.
    """
    agent = build_web_agent(model_name=model_name)

    if stream:
        console.print("[bold]Running (streaming)…[/bold]")
        for event in agent.stream({"messages": [("user", query)]}):
            # event is a dict keyed by node name -> output delta
            for node, data in event.items():
                # data may contain a list of messages; print incremental chunks
                msg_list = data.get("messages") if isinstance(data, dict) else None
                if msg_list:
                    last = msg_list[-1]
                    role = getattr(last, "type", getattr(last, "role", ""))
                    content = getattr(last, "content", str(last))
                    if content:
                        console.print(f"[dim]{node}[/dim]: {content}")
        console.print("[bold green]Done.[/bold green]")

    else:
        result = agent.invoke({"messages": [("user", query)]})
        messages = result.get("messages", [])
        if messages:
            console.print(messages[-1].content)


# if __name__ == "__main__":
#     parser = argparse.ArgumentParser(description="LangGraph web-search agent using DuckDuckGo tool")
#     parser.add_argument("--q", "--query", dest="query", required=True, help="Your question or search intent")
#     parser.add_argument("--no-stream", action="store_true", help="Disable streaming of intermediate steps")
#     parser.add_argument("--model", type=str, default=None, help="OpenAI model id, e.g., gpt-4o-mini")
#     args = parser.parse_args()

#     ask_web_agent(query=args.query, model_name=args.model, stream=not args.no_stream)
start_time = time.time()
ask_web_agent(query="what is python", model_name="gpt-4o-mini")
end_time = time.time()


print(f"Execution time: {end_time - start_time:.4f} seconds")
```
